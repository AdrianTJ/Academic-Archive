```python
import pandas
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pandas.read_csv("../IMDb_Extensive_DatasetIMDb/movies.csv")
data = pandas.DataFrame.dropna(data)
```

```python
data.info()
```

```python
data["year"].unique()
data["year"] = data["year"].astype('category').cat.codes
data["year"].unique()
```

```python
usa_income_series = pandas.Series.dropna(data['usa_gross_income'])
usa_income_series = usa_income_series[usa_income_series.astype('str').str.contains("$", regex = False)]
usa_income_series = pandas.to_numeric(pandas.Series(usa_income_series.astype(str).str.extract('(\d+)')[0]))
data["usa_gross_income"] = np.log(usa_income_series)
```

```python
# Separa la base de datos, usa la variable "quality" como variable objetivo, y
# todas las demás variables como variables explicativas
# X = data[["type","fixed acidity","volatile acidity","citric acid",
#           "residual sugar","chlorides","free sulfur dioxide",
#           "total sulfur dioxide","density","pH","sulphates","alcohol"]].to_numpy()

X = data[["duration", "avg_vote", "votes"]].to_numpy()
Y = data["usa_gross_income"].to_numpy()

print(X.shape)
print(Y.shape)
```

```python
# Revisa la documentación de sklearn.model_selection.train_test_split(), y
# encutra la manera de dividir la base de datos en tres set aleatorios. Usa:
# 70% para entrenamiento, 10% para validación, y 20% para prueba
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.1, random_state=42)

print(f"{len(x_train)} training samples")
print(f"{len(x_val)} validation samples")
print(f"{len(x_test)} test samples")
```

```python
# Usa kNN de scikit learn
from sklearn.neighbors import KNeighborsRegressor

# Crea un objeto y entrenalo
kNN_classifier = KNeighborsRegressor(n_neighbors=4)
kNN_classifier.fit(x_train, y_train)
```

```python
# Predict on the validation set
y_val_hat = kNN_classifier.predict(x_val)
print("prediction: ", y_val_hat[:20])
print("true label: ", y_val[:20])
```

```python
# Report accuracy performance on the training and validation set
print(f"Performance on the training set: {kNN_classifier.score(x_train, y_train):.3f}")
print(f"Performance on the validation set: {kNN_classifier.score(x_val, y_val):.3f}")
```
